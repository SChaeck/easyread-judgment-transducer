```python
from datasets import load_dataset, Dataset
import pandas as pd

# 데이터 로드
dataset = load_dataset('json', data_files='judgment_transducer_DPO_dataset_3.jsonl', split='train')

# 데이터셋을 pandas DataFrame으로 변환
df = dataset.to_pandas()

# 길이 기준으로 아웃라이어 제거
# 예: 길이가 1000보다 긴 데이터 포인트 제거
# 데이터 포인트의 길이를 확인할 기준은 데이터의 형식에 따라 다릅니다.
# 아래 예제에서는 'text'라는 열의 길이를 기준으로 합니다.
df['length'] = df['response_k'].apply(lambda x: len(x))  # 'text' 열의 길이
filtered_df = df[df['length'] <= 1750]  # 길이가 1000 이하인 데이터만 필터링

# DataFrame을 다시 Dataset으로 변환
filtered_dataset = Dataset.from_pandas(filtered_df)
```

```python

# 수정된 데이터셋을 파일로 저장
filtered_dataset.to_json('filtered_judgment_transducer_DPO_dataset_10k.jsonl')

# 수정된 데이터셋을 허깅페이스 허브에 업로드
filtered_dataset.push_to_hub("Suchae/judgment-transducer-DPO-dataset-10k")
```
